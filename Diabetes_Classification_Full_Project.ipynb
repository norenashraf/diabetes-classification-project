# Data Understanding



## حجم الداتا

1000 صف.
14 عمود، منهم 13 عمود إدخال وعمود هدف واحد.

## شرح كل عمود بالتفصيل

ID
رقم تسلسلي للسجل نفسه (مش للمريض). مجرد ترقيم داخلي في الملف، مفيش أي علاقة طبية بيه، فهيتشال قبل التدريب.

No_Pation
رقم ملف المريض في المستشفى أو العيادة. زي الرقم القومي بالظبط، بيميّز كل مريض عن التاني، لكن الرقم نفسه مالوش دلالة طبية، فهيتشال برضه.

Gender
جنس المريض: M يعني ذكر، أو F يعني أنثى. بعض المؤشرات الطبية زي معدلات الكرياتينين والدهون بتختلف طبيعيًا بين الرجالة والستات، فده بيفيد الموديل.

AGE
عمر المريض بالسنين. خطر الإصابة بالسكري بيزيد مع التقدم في العمر، خصوصًا بعد سن الأربعين.

Urea
نسبة اليوريا في الدم، ناتج تكسير البروتين وبيتم التخلص منه عن طريق الكلى. المعدل الطبيعي تقريبًا من 2.5 لحد 7.1. ارتفاعه بيدل على ضعف في وظائف الكلى، وهي من المضاعفات الشائعة للسكري المزمن.

Cr
الكرياتينين، ناتج تكسير عضلي وبيتصفّى بره الجسم عن طريق الكلى برضه. زيادته عن المعدل الطبيعي، تقريبًا من 60 لحد 110، بتدل على تدهور في وظائف الكلى.

HbA1c

ده أهم عمود في الداتا كلها.

بيقيس متوسط نسبة السكر في الدم على مدار آخر شهرين أو تلاتة، مش قياس لحظي زي السكر العادي. المعدل الطبيعي أقل من 5.7، من 5.7 لحد 6.5 يعتبر مرحلة قبل السكري، وأكتر من 6.5 يعتبر تشخيص مؤكد بالسكري. ده العمود اللي بيقود قرار الموديل بشكل شبه مباشر.

Chol
الكوليسترول الكلي في الدم. مرضى السكري غالبًا بيكون عندهم اضطراب في نسب الدهون كنتيجة مصاحبة للمرض.

TG
الدهون الثلاثية، نوع من الدهون بيرتفع عادة عند مرضى السكري ومقاومة الإنسولين.

HDL
الكوليسترول الجيد. بيساعد في التخلص من الكوليسترول الزائد. انخفاضه مرتبط بزيادة خطر السكري وأمراض القلب.

LDL
الكوليسترول الضار. ارتفاعه بيزوّد خطر تصلب الشرايين، وشائع عند مرضى السكري.

VLDL
نوع تالت من الكوليسترول، بينقل الدهون الثلاثية في الدم. زيادته مؤشر إضافي على اضطراب الدهون المصاحب للسكري.

BMI
مؤشر كتلة الجسم. الوزن بالكيلوجرام مقسوم على مربع الطول بالمتر. زيادة الوزن والسمنة من أكبر عوامل خطر الإصابة بالسكري من النوع الثاني.

CLASS

ده عمود الهدف.

التشخيص النهائي للمريض:

حرف N معناه غير مصاب.

حرف P معناه في مرحلة ما قبل السكري.

حرف Y معناه مصاب بالسكري.

## الهدف من المشروع

بناء موديل تصنيف يقدر يتوقّع التشخيص بالاعتماد على باقي الأعمدة، وده ممكن يُستخدم كأداة مساعدة للأطباء في الفرز الأولي لحالات السكري.

## Check Outliers

```python
def detect_outliers_iqr(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[col] < lower) | (data[col] > upper)].index

for col in df.select_dtypes(include = 'number').columns :        
    outliers = detect_outliers_iqr(df, col)
    print(col)
    print('Number of outliers:', len(outliers))
    print('-'*50)
```

#  Feature Engineering

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal'
    elif bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'

df['BMI_Category'] = df['BMI'].apply(bmi_category)
df['BMI_Category'].value_counts()
```

# Data Preprocessing

### Split Data into Feature and Target

```python
x = df.drop('CLASS',axis = 1)
y = df['CLASS']
```

###  Target Encoding

#### N=0 , P=1 , Y=2


```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
dict(zip(le.classes_, range(len(le.classes_))))     
```

###  Numerical Handling

```python
num_cols = x.select_dtypes(include = 'number').columns
num_cols
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
num_pipeline = Pipeline([('scaler', RobustScaler())])
num_pipeline
```

### Categorical Handling

```python
from sklearn.preprocessing import OneHotEncoder

cat_cols = ['Gender', 'BMI_Category']
cat_pipeline = Pipeline([('encoder', OneHotEncoder(drop='first', handle_unknown='ignore'))])
cat_pipeline
```

```python
from sklearn.compose import ColumnTransformer
Preprocessing = ColumnTransformer([('num',num_pipeline,num_cols),('cat', cat_pipeline, cat_cols)])
Preprocessing
```

###  Train/Test Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
x_train.shape, x_test.shape
```

## Machine Learning

### Check All Models without changing in hyperparameters

هنستخدم مقياسي الدقة والـ F1 Macro مع بعض، لإن مقياس الدقة لوحده ممكن يكون مش صح اوي مع الكلاسات اللي مش متوازنه. مقياس الـ F1 Macro بيدي وزن متساوي لكل كلاس بغض النظر عن حجمه.

```python
from sklearn.model_selection import cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier


models = [
    
    ('Logistic Regression', LogisticRegression(max_iter=1000, class_weight='balanced')),
    ('KNN', KNeighborsClassifier()),
    ('Decision Tree', DecisionTreeClassifier(random_state= 42, class_weight='balanced')),
    ('Random Forest', RandomForestClassifier(random_state= 42, class_weight='balanced')),
    ('Gradient Boosting', GradientBoostingClassifier(random_state= 42))
]

for model in models:

    model_pipeline = Pipeline(steps= [ ('Preprocessing', Preprocessing),
                                       ('Model', model[1])])
    
    result = cross_validate(model_pipeline, x_train, y_train, scoring= ['accuracy','f1_macro'], cv= 5, return_train_score= True, n_jobs= -1)

    print(model[0])
    print('Train Accuracy :', round(result['train_accuracy'].mean() * 100, 2))
    print('Test Accuracy  :', round(result['test_accuracy'].mean() * 100, 2))
    print('Train F1 Macro :', round(result['train_f1_macro'].mean() * 100, 2))
    print('Test F1 Macro  :', round(result['test_f1_macro'].mean() * 100, 2))
    print('-' * 50)
```

### Trying Random Forest with cross val and tuning

```python
from sklearn.model_selection import RandomizedSearchCV

rf_pipeline = Pipeline([
    ('Preprocessing', Preprocessing),
    ('Model', RandomForestClassifier(random_state=42, class_weight='balanced'))
])

rf_param_dist = {
    'Model__n_estimators': [100, 200, 300],
    'Model__max_depth': [5, 10, 15, None],
    'Model__min_samples_split': [2, 5, 10],
    'Model__min_samples_leaf': [1, 2, 4]
}

rf_search = RandomizedSearchCV(rf_pipeline, rf_param_dist, n_iter=10, cv=5, scoring='f1_macro', return_train_score= True, n_jobs=-1, verbose=1, random_state=42)
rf_search.fit(x_train, y_train)

print("Random Forest Best Score (F1 Macro):", round(rf_search.best_score_ * 100, 2))
print("Random Forest Best Params:", rf_search.best_params_)
train_mean = rf_search.cv_results_['mean_train_score'][rf_search.best_index_]
test_mean = rf_search.cv_results_['mean_test_score'][rf_search.best_index_]
print('Train F1 Macro:', round(train_mean * 100, 2))
print('Test F1 Macro :', round(test_mean * 100, 2))
```

### Trying Gradient Boosting with cross val and tuning

```python
gb_pipeline = Pipeline([
    ('Preprocessing', Preprocessing),
    ('Model', GradientBoostingClassifier(random_state=42))
])

gb_param_dist = {
    'Model__n_estimators': [100, 200, 300],
    'Model__max_depth': [2, 3, 4, 5],
    'Model__learning_rate': [0.01, 0.05, 0.1],
    'Model__subsample': [0.7, 0.8, 0.9, 1.0]
}

gb_search = RandomizedSearchCV(gb_pipeline, gb_param_dist, n_iter=10, cv=5, scoring='f1_macro', return_train_score= True, n_jobs=-1, verbose=1, random_state=42)
gb_search.fit(x_train, y_train)

print("Gradient Boosting Best Score (F1 Macro):", round(gb_search.best_score_ * 100, 2))
print("Gradient Boosting Best Params:", gb_search.best_params_)
train_mean = gb_search.cv_results_['mean_train_score'][gb_search.best_index_]
test_mean = gb_search.cv_results_['mean_test_score'][gb_search.best_index_]
print('Train F1 Macro:', round(train_mean * 100, 2))
print('Test F1 Macro :', round(test_mean * 100, 2))
```

### Final Evaluation on Test Set

```python
from sklearn.metrics import classification_report, confusion_matrix
import plotly.figure_factory as ff

best_model = rf_search.best_estimator_    

y_pred = best_model.predict(x_test)

print(classification_report(y_test, y_pred, target_names=le.classes_))
```

```python
cm = confusion_matrix(y_test, y_pred)
fig = px.imshow(cm, text_auto=True, x=le.classes_, y=le.classes_,
                 labels=dict(x="Predicted", y="Actual", color="Count"),
                 title="Confusion Matrix")
fig.show()
```

#  Last step before Deployment: save pipeline in pkl file

```python
import joblib

final_pipeline = Pipeline([
    ('Preprocessing', Preprocessing),
    ('Model', RandomForestClassifier(
        n_estimators=rf_search.best_params_['Model__n_estimators'],
        max_depth=rf_search.best_params_['Model__max_depth'],
        min_samples_split=rf_search.best_params_['Model__min_samples_split'],
        min_samples_leaf=rf_search.best_params_['Model__min_samples_leaf'],
        class_weight='balanced',
        random_state=42
    ))
])

final_pipeline.fit(x, y)     

joblib.dump(final_pipeline, 'final_diabetes_classification_model.pkl')
joblib.dump(le, 'label_encoder.pkl')       
print("Final model saved as 'final_diabetes_classification_model.pkl'")
```

### Deployment (Streamlit App Sketch)

```python
%%writefile Diabetes_Prediction_deployment.py

import pandas as pd
import numpy as np
import plotly.express as px
import streamlit as st
import joblib

st.set_page_config(layout='wide', page_title='Diabetes Diagnosis Classifier', page_icon='🏥')

# ---------- Palette (exact colors provided) ----------
BG = '#FFF6FB'      # page background - very light pink
PINK = '#FFD6E8'
SKY = '#7FD6FF'
MINT = '#6FE3B4'
NAVY = '#2B4A66'    # sidebar / headers / dark accent

st.markdown(f"""
<style>
.stApp {{ background-color: {BG}; }}
section[data-testid="stSidebar"] {{ background-color: {NAVY}; }}
section[data-testid="stSidebar"] * {{ color: {BG} !important; }}
h1, h2, h3 {{ color: {NAVY}; }}
.stButton>button {{
    background-color: {NAVY}; color: white; border-radius: 8px; border: none;
}}
div[data-baseweb="select"] > div {{ border-color: {NAVY} !important; background-color: white; }}
.block-container {{ padding-top: 2rem; }}
[data-testid="stMetric"] {{
    background-color: white; border: 1px solid {PINK}; border-left: 6px solid {NAVY};
    border-radius: 10px; padding: 10px 14px;
}}
</style>
""", unsafe_allow_html=True)

# CLASS colors mapped from the same palette
CLASS_COLORS = {'N': MINT, 'P': SKY, 'Y': NAVY}
SEQ_SCALE = [[0, BG], [0.5, SKY], [1, NAVY]]

st.sidebar.title("⚕️ Navigation")
page = st.sidebar.radio("", ["Dataset information", "Data analysis", "Prediction"])

@st.cache_data
def load_data():
    return pd.read_csv('diabetes_data_cleaned.csv')

@st.cache_resource
def load_model():
    model = joblib.load('final_diabetes_classification_model.pkl')
    le = joblib.load('label_encoder.pkl')
    return model, le

df = load_data()
model, le = load_model()

PLOT_LAYOUT = dict(plot_bgcolor='white', paper_bgcolor='white', font_color=NAVY)

if page == "Dataset information":
    st.title("💉 Diabetes Diagnosis Dataset")

    c1, c2, c3, c4 = st.columns(4)
    c1.metric("إجمالي المرضى (Total)", len(df))
    c2.metric("غير مصاب (N)", int((df['CLASS'] == 'N').sum()))
    c3.metric("قبل السكري (P)", int((df['CLASS'] == 'P').sum()))
    c4.metric("مصاب (Y)", int((df['CLASS'] == 'Y').sum()))

    st.subheader("Dataset Preview")
    st.dataframe(df, use_container_width=True)

elif page == "Data analysis":
    st.title("📊 Data Analysis")

    chart_choice = st.selectbox(
        'اختاري الرسم البياني (Chart)',
        [
            'عدد المرضى في كل تشخيص (Diagnosis Counts)',
            'متوسط مؤشر كتلة الجسم لكل تشخيص (Mean BMI per Class)',
            'متوسط السكر التراكمي لكل تشخيص (Mean HbA1c per Class)',
            'توزيع التشخيص حسب الجنس (Diagnosis by Gender)',
            'توزيع الأعمار حسب التشخيص (Age Distribution by Class)',
            'مقارنة متوسط قيم الدهون لكل تشخيص (Lipid Comparison)'
        ]
    )

    if chart_choice.startswith('عدد المرضى'):
        counts = df['CLASS'].value_counts().reset_index()
        counts.columns = ['CLASS', 'count']
        fig = px.pie(
            counts, names='CLASS', values='count', color='CLASS',
            color_discrete_map=CLASS_COLORS,
            title='عدد المرضى في كل تشخيص', hole=0.45
        )
        fig.update_layout(**PLOT_LAYOUT)
        st.plotly_chart(fig, use_container_width=True)

    elif chart_choice.startswith('متوسط مؤشر'):
        df_grouped = df.groupby('CLASS')['BMI'].mean().sort_values(ascending=False).reset_index()
        fig = px.bar(
            df_grouped, x='CLASS', y='BMI', color='CLASS',
            color_discrete_map=CLASS_COLORS,
            title='متوسط مؤشر كتلة الجسم لكل تشخيص', text_auto='.2s'
        )
        fig.update_layout(**PLOT_LAYOUT)
        st.plotly_chart(fig, use_container_width=True)

    elif chart_choice.startswith('متوسط السكر'):
        df_grouped = df.groupby('CLASS')['HbA1c'].mean().sort_values(ascending=False).reset_index()
        fig = px.bar(
            df_grouped, x='CLASS', y='HbA1c', color='CLASS',
            color_discrete_map=CLASS_COLORS,
            title='متوسط السكر التراكمي لكل تشخيص', text_auto='.2s'
        )
        fig.update_layout(**PLOT_LAYOUT)
        st.plotly_chart(fig, use_container_width=True)

    elif chart_choice.startswith('توزيع التشخيص'):
        gender_class = df.groupby(['Gender', 'CLASS']).size().reset_index(name='count')
        fig = px.bar(
            gender_class, x='Gender', y='count', color='CLASS', barmode='group',
            color_discrete_map=CLASS_COLORS,
            title='توزيع التشخيص حسب الجنس'
        )
        fig.update_layout(**PLOT_LAYOUT)
        st.plotly_chart(fig, use_container_width=True)

    elif chart_choice.startswith('توزيع الأعمار'):
        fig = px.histogram(
            df, x='AGE', color='CLASS', color_discrete_map=CLASS_COLORS,
            title='توزيع الأعمار حسب التشخيص', barmode='overlay', opacity=0.75
        )
        fig.update_layout(**PLOT_LAYOUT)
        st.plotly_chart(fig, use_container_width=True)

    else:
        lab_cols = ['Chol', 'TG', 'HDL', 'LDL', 'VLDL']
        melted = df.melt(id_vars='CLASS', value_vars=lab_cols, var_name='Lab Test', value_name='Value')
        grouped = melted.groupby(['CLASS', 'Lab Test'])['Value'].mean().reset_index()
        fig = px.bar(
            grouped, x='Lab Test', y='Value', color='CLASS', barmode='group',
            color_discrete_map=CLASS_COLORS,
            title='مقارنة متوسط قيم الدهون لكل تشخيص'
        )
        fig.update_layout(**PLOT_LAYOUT)
        st.plotly_chart(fig, use_container_width=True)



elif page == "Prediction":
    st.title("🔮 Predict Diabetes Diagnosis")

    gender = st.selectbox('الجنس (Gender)', ['M', 'F'])
    age = st.number_input('العمر  (Age)', 1, 100, 40)
    urea = st.number_input('نسبة اليوريا في الدم (Urea)', 0.0, 50.0, 5.0)
    cr = st.number_input('الكرياتينين (Cr)', 0, 500, 50)
    hba1c = st.number_input('السكر التراكمي (HbA1c %)', 0.0, 20.0, 5.5)
    chol = st.number_input('الكوليسترول الكلي (Chol)', 0.0, 15.0, 4.5)
    tg = st.number_input('الدهون الثلاثية (TG)', 0.0, 15.0, 1.5)
    hdl = st.number_input('الكوليسترول الجيد (HDL)', 0.0, 10.0, 1.2)
    ldl = st.number_input('الكوليسترول الضار (LDL)', 0.0, 15.0, 2.5)
    vldl = st.number_input('كوليسترول منخفض الكثافة جدًا (VLDL)', 0.0, 40.0, 1.0)
    bmi = st.number_input('مؤشر كتلة الجسم (BMI)', 10.0, 50.0, 25.0)

    def bmi_category(bmi):
        if bmi < 18.5: return 'Underweight'
        elif bmi < 25: return 'Normal'
        elif bmi < 30: return 'Overweight'
        else: return 'Obese'

    if st.button('Predict'):
        input_df = pd.DataFrame([{
            'Gender': gender, 'AGE': age, 'Urea': urea, 'Cr': cr, 'HbA1c': hba1c,
            'Chol': chol, 'TG': tg, 'HDL': hdl, 'LDL': ldl, 'VLDL': vldl, 'BMI': bmi,
            'BMI_Category': bmi_category(bmi)
        }])
        pred = model.predict(input_df)
        pred_label = le.inverse_transform(pred)[0]
        st.success(f"Predicted Class: {pred_label}")
```
